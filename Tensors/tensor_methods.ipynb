### Shape Manipluation

#### 1.Reshape

```python
tensor = torch.tensor([10,20,30,40,50,60])
tensor
```

```python
print(f"Shape of Tensor : {tensor.shape}")
print(f"ndim : {tensor.ndim}")
```

```python
tensor_reshaped = tensor.reshape(6,1)
print(f"tensor : {tensor_reshaped}")
print(f"Shape of Tensor : {tensor_reshaped.shape}")
print(f"ndim : {tensor_reshaped.ndim}")
```

#### Flatten

```python
# Create the 2d tensor
tensor = torch.tensor([[10,20],[30,40],[50,60]])
tensor
```

```python
tensor_flttened = tensor.flatten()
print(f"tensor : {tensor_flttened}")
print(f"Shape of Tensor : {tensor_flttened.shape}")
print(f"ndim : {tensor_flttened.ndim}")
```

#### Transpose : t()

```python
print("Befor Transpose : ")
tensor = torch.tensor([[10,20],[30,40],[50,60]])
print(f"tensor : {tensor}")
print(f"Shape of Tensor : {tensor.shape}")
print(f"ndim : {tensor.ndim}")
tensor_transposed = tensor.t()
print("\nAfter Transpose : ")
print(f"tensor : {tensor_transposed}")
print(f"Shape of Tensor : {tensor_transposed.shape}")
print(f"ndim : {tensor_transposed.ndim}")
```
