```python
import torch
import numpy as np
```

### Matrix Multiplication

```python
t1 = torch.tensor([[10,20,30,40,50,60]])
t2 = torch.tensor([[1,2,3,4,5,6]])
t2=t2.reshape(6,1)
print(t1)
print(f"Shape of t1 : {t1.shape}, t2 : {t2.shape}")
```

#### Matrix multiplication

```python
t1@t2
```

```python
t1.matmul(t2)
```

```python
t1.mm(t2)  ## For 2D
```

```python
vec = torch.tensor([10,20,30,40,50,60])
res = t1.mv(vec)
print(res)
```

#### Dot Product 

```python
res = torch.dot(t1.squeeze(),t2.squeeze())
res
```

#### Inner Product : shapes should same

```python
t2=t2.reshape(1,6)
t1.inner(t2)
```

#### Outer Product : 

```python
t1=t1.squeeze()
t2=t2.squeeze()
print(torch.outer(t1,t2))
```

#### Trace

```python
t3 = torch.tensor([[1,2,3],[1,2,3],[1,2,3]])
t3.trace()
```

```python
t3.diag()
```

#### Matrix inverse

```python
t3 = torch.tensor([[1.0, 2.0], 
                  [3.0, 4.0]])

# Calculate inverse
inv_t3 = torch.inverse(t3)

print(inv_t3)
```

### Determinant

```python
t3.det()
```

#### Eigan Values

```python
eigenvalues, eigenvectors = torch.linalg.eig(t3)

print("Eigenvalues:\n", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```
