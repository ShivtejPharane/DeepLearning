```python
import torch
import numpy as np
```

### Histogram Count

```python
x = torch.tensor([1,2,2,3,4,5,6,7,8,9],dtype=float)
hist = torch.histc(x,bins=4,min=1,max=9)
hist
```

### Count Of Integers

```python
x1 = torch.tensor(x,dtype=int)
x1.bincount()
```

```python
x = torch.tensor([0,1,1,2])
weights = torch.tensor([1.0,2.0,3.0,4.0])

print(torch.bincount(x, weights=weights))
```

#### torch.coeff()

```python
x=torch.tensor([
    [1.,2.,3.,4.],
    [4.,3.,2.,1.]
])
print(torch.corrcoef(x))
```

#### Covariance matrix

```python
x=  torch.tensor([
    [1.,2.,3.],
    [2.,4.,6.]
])
print(torch.cov(x))
```
