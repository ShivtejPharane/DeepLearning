### Text preprocessing

```python
from nltk import word_tokenize
all_words = []
for review in reviews:
    words = word_tokenize(review)
    all_words.extend(words)
print(len(all_words))
```

```python
lower_case=[]
for word in all_words:
    lower_case.append(word.lower())
```

```python
print(len(lower_case))
```

```python
from nltk import pos_tag
pos_tags = pos_tag(lower_case)
```

```python
from nltk.stem import WordNetLemmatizer
lemmetizer = WordNetLemmatizer()
lemmatized_words = []
for word, pos in pos_tags:
    tag='n'
    if pos.startswith('N'):
        tag = 'n'
    if pos.startswith('V'):
        tag = 'v'
    if pos.startswith('A'):
        tag = 'a'
    base_form = lemmetizer.lemmatize(word,pos=tag)
    lemmatized_words.append(base_form)
    print(base_form)
```

```python
lemmatized_words = [word for word in lemmatized_words if len(word)!=1]
```

### Build the vocabulary

```python
# set the max length of every word in vocabulary
MAX_LENGTH = 5
vocabulary = {
    # we will keep the words with fixed length
    # if the word is smaller than the fixed length, we will add padding
    "<PAD>":0,
    "<UNK>":1
}
```

```python
# split all the rwviews into tokans(words)
for word in lemmatized_words:
        if word not in vocabulary:
            vocabulary[word] = len(vocabulary)
```

```python
vocabulary
```

### Encode all the reviews

```python
def encode(words):
    token_ids = []

    for word in words:
        token_id = vocabulary.get(word, vocabulary["<UNK>"])
        token_ids.append(token_id)

    return token_ids
```

```python
encode(reviews[1].split())
```

```python
encoded_reviews = [encode(review) for review in reviews]
```
