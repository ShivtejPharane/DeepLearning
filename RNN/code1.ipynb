### Import the Required packages

```python
import numpy as np
import torch
```

### get the Data

```python
x = torch.tensor([
    [
        [10.0,20.0,30.0,40.0,50.0]
    ]
])
```

### Create RNN Layer

```python
rnn = torch.nn.RNN(
    input_size = 5,
    hidden_size=5,
    batch_first=True
)
```

### Pass the Data to rnn

```python
output,hidden = rnn(x)
```
