### DataLoader Demo

### Problem Statement
- instead of loading whole dataset , Load it batch by batch

```python
import numpy as np
import pandas as pd
import torch
```

### Genrate the Dummy Data

```python
#number of rows
number_of_rows = 10000
#number of featuers
number_of_featuers = 20
```

```python
# genrate the x Data
x = torch.randn(number_of_rows,number_of_featuers)
x
```

```python
x.shape
```

```python
# genrate the y data
y = torch.randint(0,2,(number_of_rows,))
y
```

```python
y.shape
```

### Create the Tensor Dataset

```python
from torch.utils.data import TensorDataset

# create the dataset  
dataset = TensorDataset(x,y)
```

```python
dataset.tensors
```

### Create a data Loader

```python
from torch.utils.data import DataLoader
# create the loader to load dataset in Batches
loader = DataLoader(dataset=dataset,batch_size=128,shuffle=True)
```

```python
# print the loader information
print(f"total number of records : {len(dataset)}")
print(f"batch size of loader : {loader.batch_size} ")
print(f"Total number of Loaders : {len(loader)}")
```

### Iterate over  the dataloader to read the data batch by batch

```python
for (batch_index,(features,labels)) in enumerate(loader):
    print(f"Batch No : {batch_index+1}")
    print(f"data.shape : {features.shape} ")
```
