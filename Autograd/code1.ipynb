```python
import numpy as np
import torch
```

### Create the Data

```python
# Scalar Tensor
x=torch.tensor(10,requires_grad=False)
print(x)
```

### Mannual Cal

```python
# cal another tensor using x
y= x**2
print(y)
```

```python
# cal derivative of y w.r.t x
# the darivative of x^2 -> 2x
darivative_of_y=2*x
print(darivative_of_y)
```

### Auto Grad

```python
# create the tensor 
a = torch.tensor(10.0,requires_grad=True)
print(a)
print(f"gradient stored in a = {a.grad}")
```

```python
# cal another tensor using a
b = a**2
print(f"b : {b}")
```

```python
# cal gradient Automatically
b.backward()
```

```python
# check the gradients 
print(f"gradient stored in a = {a.grad}")
```

```python
# check the grad function
print(b.grad_fn)
```

```python
c=a**3
c.backward()
```

```python
print(a.grad)
```

### Note 

- Since a.grad alredy was having the value 20,
- The final answer is 320 -> 300+20=320 not 300
- This Called the Gradient accumlatuion

```python
# if u want to clear the privious gradients
a.grad = torch.tensor(0.0)
```
