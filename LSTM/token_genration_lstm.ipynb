### Text preprocessing

```python
tokens = data.split()
```

```python
print(tokens)
```

```python
vocabulary = sorted(set(tokens))
print(vocabulary)
```

```python
# word to index:
word_to_index={word:index for index,word in enumerate(vocabulary)}
print(word_to_index)
```

```python
# dictonary to store the indexes with their respective word
index_to_word = {index:word for index,word in enumerate(vocabulary)}
print(index_to_word)
```

### Gentate training data

```python
# define the sequence length
# every input in training data will be have 3 tokens and 4th will be output
SEQUENCE_LENGTH = 3
```

```python
# build the input and outpu
x=[]
y=[]
```

```python
for index in range(len(tokens)-SEQUENCE_LENGTH):
    # genrate the input sequence
    input_sequence = tokens[index:index+ SEQUENCE_LENGTH]

    #append the corresponding index positions(numric format)
    x.append([word_to_index[word] for word in input_sequence])
    
    # genrate the output sequence    
    output = tokens[index+SEQUENCE_LENGTH]

    # append the corresponding index position of output
    y.append(word_to_index[output])
    #print(f"input : {input_sequence} output : {output} ")
```

```python
print(x)
```

```python
print(y)
```

```python
# create the tensors from the list
x = torch.tensor(x)
y = torch.tensor(y)
```

### Define the Model

```python
class NextWordPrediction(torch.nn.Module):
    def __init__(self,vocabulary_size,embbeding_dim,hidden_dim):
        #initalize the parent class
        super().__init__()

        # create the embedding layer
        self.embbeding = torch.nn.Embedding(num_embeddings=vocabulary_size,embedding_dim=embbeding_dim)
        # add the LSTM layer for learning the from the data
        self.lstm = torch.nn.LSTM(
            input_size = embbeding_dim,
            hidden_size = hidden_dim,
            batch_first = True 
        )

        # add a Linear Layer for genrating the next token
        self.linear = torch.nn.Linear(hidden_dim,vocabulary_size)
    def forward(self,x):
        x = self.embbeding(x)

        # pass the embedding to the LSTM Layer
        output,(hidden,cell) = self.lstm(x)

        x = hidden[-1]
        # pass the reshaped hidden state to the linera layer
        x = self.linear(x)
        return x
```

```python
device = ''
if torch.cuda.is_available():
    device='cuda'
else:
    device='cpu'
```

```python
# create the model 
model = NextWordPrediction(
    vocabulary_size=len(vocabulary),
    embbeding_dim=16,
    hidden_dim=32
)
model = model.to(device)
```

### define the Hyper parmeters

```python
loss_function = torch.nn.CrossEntropyLoss()

epochs = 500

optimizer = torch.optim.Adam(model.parameters(),lr=0.01)
```

### training loop

```python
model.train()
x = x.to(device)
y = y.to(device)
for epoch in range(epochs):
    # prediction
    predicted = model(x)

    loss = loss_function(predicted,y)
    
    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    if (epoch+1)%50==0:
        print(f"epoch : {epoch}  loss : {loss.item()}")
```

### predict the next tocken on unseen data

```python
def predict_text(sequence):
    sequence = sequence.lower()
    words = sequence.split()
    
    if len(words) != SEQUENCE_LENGTH:
        print(f"Exactly {SEQUENCE_LENGTH} number of words required")
        return None

    input_tokens = [word_to_index[word] for word in words]
    input_tensor = torch.tensor(input_tokens).unsqueeze(0).to(device)
    
    model.eval()
    with torch.no_grad():
        output = model(input_tensor)
        predicted_index = torch.argmax(output, dim=1).item()
        
        # Return the predicted word instead of just printing it
        return index_to_word[predicted_index]
```
