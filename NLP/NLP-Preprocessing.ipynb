### Load the data

```python
data = """The figures are based on data from GlobalPetrolPrices, which tracks fuel prices in 170 countries and territories. Petrol prices in Myanmar rose the most, increasing by 56 percent from $0.77 per litre of 95-octane fuel on February 23 to $1.20 on August 17. Bhutan recorded the next-largest increase at 55 percent, followed by Cuba at 51 percent, the UAE at 50 percent and 48 percent in Nigeria. In 25 other countries, most of them oil producers with heavily subsidised fuel, prices have either remained unchanged or fallen by single digits."""
```

### Sentence Segmentation

```python
sentences = nltk.tokenize.sent_tokenize(data)
print(len(sentences))
print(sentences)
```

### Word Tokenization

```python

all_words = []
for sentence in sentences:
    words = nltk.tokenize.word_tokenize(sentence)
    all_words.extend(words)

print(all_words)
```

```python
print(len(all_words))
```

```python
all_words = []
all_words = nltk.tokenize.word_tokenize(data)
print(len(all_words))
print(all_words)
```

### POS tagging

```python
import nltk
from nltk.tag import pos_tag

nltk.download("averaged_perceptron_tagger_eng")

pos_tags = pos_tag(all_words)
```

### Stemming

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ['better','running','apple','ran']
for word in words:
    base_form = stemmer.stem(word)
    print(base_form)
```

### Lemmatization

```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")

words_ = [("better", "a"), ("running", "v"), ("apples", "n"), ("ran", "v"),('best',"a")]
lemmatizer = WordNetLemmatizer()

for word, pos in words:
    base_form = lemmatizer.lemmatize(word, pos=pos)
    print(base_form)
```

```python
pos_tags
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = []
for word, pos_tag in pos_tags:
    tag = 'n'
    if pos_tag.startswith('N'):
        tag = 'n'
    elif pos_tag.startswith('V'):
        tag = 'v'
    elif pos_tag.startswith('A'):
        tag = 'a'
    base_form = lemmatizer.lemmatize(word, pos=tag)
    lemmatized_words.append(base_form)
    print(f"{word} : {base_form}")
```

### Lower casing

```python
lowercasing = [word.lower() for word in lemmatized_words]
```

```python
lowercasing
```

### Removing symbols 

```python
symbols = ['.',',','"',"'",'!',':',';','$']

all_words_without_symbol = [word for word in lowercasing if word not in symbols]
all_words_without_symbol
```

```python
print(len(all_words_without_symbol))
```

### Remove the Stop Words

```python
from nltk.corpus import stopwords

stop_words = stopwords.words('english')
print(len(stop_words))
print(stop_words)
```

```python
all_words_without_stopwards = [word for word in all_words_without_symbol if word not in stop_words]
print(len(all_words_without_stopwards))
print(all_words_without_stopwards)
```
