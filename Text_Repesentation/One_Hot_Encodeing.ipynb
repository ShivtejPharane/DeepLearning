```python
import numpy as np
```

### Load the Data

```python
sentence = "New York City Mayor Zohran Mamdani says he opposes a right-wing Hindu nationalist event scheduled for next week at Madison Square Garden. Mamdani says the group, which is closely linked to India’s ruling BJP party, has an ‘exclusionary vision’."
```

```python
sentence = sentence.lower() 
print(sentence)
```

```python
from nltk.tokenize import word_tokenize
words = word_tokenize(sentence)
print(words)
```

```python
lowercase_words = [word.lower() for word in words]
print(lowercase_words)
```

```python
special_characters = ['"',"'",".",",",";"]
lowercase_words = [word for word in lowercase_words if word not in special_characters and len(word)>1]
lowercase_words
```

### Build the Vocabulary

```python
vocabulary = sorted(set(lowercase_words))
print(vocabulary)
```

### assign an index to all words

```python
word_to_index = {}
for index,word in enumerate(vocabulary):
    word_to_index[word] = index

print(word_to_index)
```

### Create the One Hot Encoding

```python
word_vectors ={}
for word in vocabulary:
    vector = np.zeros(len(vocabulary),dtype=np.int8)

    position = word_to_index[word]
    vector[position] = 1
    word_vectors[word] = vector
print(word_vectors)
```

```python
for word in word_vectors:
    print(f"{word} = {word_vectors[word]}")
```
