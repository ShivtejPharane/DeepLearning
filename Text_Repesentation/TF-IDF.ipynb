```python
import numpy as np
```

### Load the Data

```python
documents = [
    "I  love python",
    "I love machine learning",
    "I love nlp"
]
```

### Text preprocessing

```python
words = [document.lower().split() for document in documents]
words
```

```python
all_words = []
for word_list in words:
    for word in word_list:
        all_words.append(word)
```

### Build the vocabulary

```python
vocabulary = sorted(set(all_words))
vocabulary
```

```python
print(len(vocabulary))
```

### cal TF

```python
# collect the frewuency of each term wrt every document
document_frequency = {}
```

```python
for word in vocabulary:
    count=0
    for document in documents:
        doc_words = document.lower().split()
        if word in doc_words:
            count+=1
    document_frequency[word] = count
```

```python
document_frequency
```

### Calculate the term frequency

```python
term_frequency = {}
for word in vocabulary:
    for document in documents:
        document_words = document.lower().split()
    term_frequency[word] = document_frequency[word]/len(document_words)
```

```python
print(term_frequency)
```

### Cal IDF

```python
import math
inverse_document_frequency = {}

for word in vocabulary:
    inverse_document_frequency[word] = math.log(len(documents)/document_frequency[word])
```

```python
inverse_document_frequency
```

### Cal TF-IDF

```python
tf_idf = {}
for word in vocabulary:
    # tf-idf = term_frequency * inverse_document_frequency
    tf_idf[word] = term_frequency[word]*inverse_document_frequency[word]
```

```python
tf_idf
```

### encode the documents

```python
encoded_documents = []

for document in documents:
    documents_words = document.lower().split()
    encoded_document = np.zeros(len(document_words))
    for index,word in enumerate(document_words):
        encoded_document[index] = tf_idf[word]

    encoded_documents.append(encoded_document)
```
