```python
import numpy as np
```

```python
sentence = "I really love ml I really love nlp I really hate ml I really hate nlp"
```

### Text Preprocesing

```python
from nltk.tokenize import word_tokenize
words = word_tokenize(sentence)
print(words)
```

```python
all_words = [word.lower() for word in words]
```

```python
all_words
```

```python
def genrate_ngram(words,n):
    grams = []
    for i in range(len(words)-n+1):
        grams.append(tuple(words[i:i+n]))
    changed_gram = []
    for gram in grams:
        if gram not in changed_gram:
            changed_gram.append(gram)
    return changed_gram
```

```python
# Genrate the Unigrams
vocabulary_for_unigram = genrate_ngram(all_words,1)
print(vocabulary_for_unigram)
```

```python
# Genrate the Bigrams
vocabulary_for_bigram = list(set(genrate_ngram(all_words,2)))
print(vocabulary_for_bigram)
```

```python
vocabulary_for_bigram = set(genrate_ngram(all_words,2))
print(len(vocabulary_for_bigram))
```

```python
# Genrate the Trigrams
vocabulary_for_trigram = genrate_ngram(all_words,3)
print(vocabulary_for_trigram)
```

```python
def bow_conversion(vocabulary):
    bow_matrix = []

    for doc in vocabulary:
        vector = np.zeros(len(vocabulary), dtype=np.int8)
        if doc in vocabulary:
            index = vocabulary.index(doc)
            vector[index] += 1

        bow_matrix.append(vector)
    bow_matrix = np.array(bow_matrix)
    for index,document in enumerate(vocabulary):
        print(f"{document} : {bow_matrix[index]}")
```

```python
bow_conversion(vocabulary_for_trigram)
```
