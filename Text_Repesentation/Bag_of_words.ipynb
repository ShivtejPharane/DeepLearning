```python
import numpy as np
```

```python
documents = [
    "I love Python Python",
    "Python is very Simple",
    "I love NLP",
    "NLP use BoW technique"
]
```

```python
from nltk.tokenize import word_tokenize
all_words = []
for sentence in documents:
    words = word_tokenize(sentence)
    all_words.extend(words)
```

```python
lowercase_words = [word.lower() for word in all_words]
```

```python
vocabulary = sorted(set(lowercase_words))
vocabulary
```

### Build the Bag of words Matrix

```python
from nltk.tokenize import word_tokenize

bow_matrix = []
for doc in documents:
    vector = np.zeros(len(vocabulary), dtype=np.int8)
    words = word_tokenize(doc)
    for word in words:
        word_lower = word.lower()
        if word_lower in vocabulary:
            index = vocabulary.index(word_lower)
            vector[index] += 1
    bow_matrix.append(vector) 
    
bow_matrix = np.array(bow_matrix)
```

```python
for index,document in enumerate(documents):
    print(f"{document:<25} = {bow_matrix[index]}")
```
